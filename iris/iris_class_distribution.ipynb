# Iris Class Distribution

This notebook counts samples in the Iris dataset, visualizes class frequency with a bar chart, and checks whether the classes are balanced.

## 1. Load the Flower Dataset

Import the libraries and load the Iris dataset from the local `iris/iris.data` file.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

columns = ["sepal_length", "sepal_width", "petal_length", "petal_width", "class"]
iris = pd.read_csv("iris/iris.data", header=None, names=columns)

iris.head()
```

## 2. Count Samples per Class

Count the number of samples in each flower class.

```python
class_counts = iris["class"].value_counts().sort_index()
class_counts
```

## 3. Create a Bar Chart of Class Counts

Visualize the class counts with a bar chart.

```python
plt.figure(figsize=(8, 5))
ax = plt.bar(class_counts.index, class_counts.values, color="#2a9d8f")

for bar in ax:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2, height + 0.5, f"{int(height)}", ha="center", va="bottom")

plt.title("Iris Flower Class Counts")
plt.xlabel("Class")
plt.ylabel("Number of Samples")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

## 4. Check Whether Classes Are Balanced

Compare the class counts to determine whether the dataset is balanced.

```python
total_samples = len(iris)
class_proportions = (class_counts / total_samples * 100).round(1)

print("Class counts:\n", class_counts)
print("\nClass proportions (%):\n", class_proportions)

balanced = class_counts.nunique() == 1
print("\nAre the classes balanced?", "Yes" if balanced else "No")
```

## 5. Summarize Visualization Insights

The bar chart shows that each flower class has 50 samples. This means the Iris dataset is perfectly balanced, which is helpful for classification because no class is overrepresented.
